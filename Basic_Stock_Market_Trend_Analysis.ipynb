## 2. Import libraries


```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
```

## 3. Download stock price data

We will use Apple (`AAPL`) as an example. You can change it to `MSFT`, `GOOGL`, `RELIANCE.NS`, or `TCS.NS`.


```python
stock = yf.download('AAPL', start='2023-01-01', auto_adjust=True, progress=False)

if stock.empty:
    print('No data found. Check your internet connection or stock symbol.')
else:
    stock = stock[['Close']].dropna()
    display(stock.head())
```

## 4. Clean and prepare the data

Remove missing closing prices and calculate moving averages.


```python
stock = stock.dropna()
stock['20-Day Average'] = stock['Close'].rolling(20).mean()
stock['50-Day Average'] = stock['Close'].rolling(50).mean()

display(stock.tail())
```

## 5. Plot the stock price


```python
plt.figure(figsize=(10, 5))
plt.plot(stock['Close'], label='Closing Price')
plt.plot(stock['20-Day Average'], label='20-Day Average')
plt.plot(stock['50-Day Average'], label='50-Day Average')
plt.title('Apple Stock Price Trend')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.show()
```

## 6. Show simple summary


```python
print('First closing price:', round(stock['Close'].iloc[0], 2))
print('Latest closing price:', round(stock['Close'].iloc[-1], 2))
print('Highest closing price:', round(stock['Close'].max(), 2))
print('Lowest closing price:', round(stock['Close'].min(), 2))
```

## Conclusion

In this project, we downloaded historical stock prices, removed missing values, calculated 20-day and 50-day moving averages, and plotted the price trend. Moving averages help summarize past price movements, but they do not guarantee future performance.
